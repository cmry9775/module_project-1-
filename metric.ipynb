```python
# Regression? Hate English
import pandas as pd
import numpy as np
from catboost import CatBoostRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

df = pd.read_csv('feature.csv')

target = 'visitors'
X = df.drop(columns=target)
y = df[target]

split_idx = int(len(df) * 0.8)

X_test = X.iloc[split_idx:]
y_test = y.iloc[split_idx:]

# 평균 Baseline은 앞 80%의 평균 사용
train_mean = y.iloc[:split_idx].mean()
mean_pred = np.full(len(y_test), train_mean)


# 저장된 CatBoost 모델
model = CatBoostRegressor()
model.load_model('visitor_estimation_model.cbm')
model_pred = model.predict(X_test)

# 평가
baseline_mae = mean_absolute_error(y_test, mean_pred)
model_mae = mean_absolute_error(y_test, model_pred)

baseline_rmse = np.sqrt(mean_squared_error(y_test, mean_pred))
model_rmse = np.sqrt(mean_squared_error(y_test, model_pred))

baseline_r2 = r2_score(y_test, mean_pred)
model_r2 = r2_score(y_test, model_pred)

print(f'테스트 데이터 수: {len(y_test)}건')
print(f'평균 예측값: {train_mean:,.1f}명\n')

print('[평균 Baseline]')
print(f'MAE : {baseline_mae:,.1f}')
print(f'RMSE: {baseline_rmse:,.1f}')
print(f'R²  : {baseline_r2:.3f}')

print('\n[CatBoost]')
print(f'MAE : {model_mae:,.1f}')
print(f'RMSE: {model_rmse:,.1f}')
print(f'R²  : {model_r2:.3f}')

print(f'\nMAE 개선율: {(baseline_mae - model_mae) / baseline_mae * 100:.1f}%')
```
